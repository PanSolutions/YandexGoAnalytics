```python
from src.core.infra.connection.databricks_connector import DatabricksConnectionService

with DatabricksConnectionService(app_name="yandex-go-analytics") as conn:
    spark = conn.spark

    print(f"Spark Version : {spark.version}")

    df = spark.range(10)
    df.show()

    spark.sql("SHOW CATALOGS").show()
```

```python
from pyspark.sql import functions as F

df = (
    spark.range(0, 1000)
    .withColumn("bucket", F.col("id") % 5)
    .groupBy("bucket")
    .agg(
        F.count("id").alias("total_rows"), F.sum("id").alias("sum_ids"), F.avg("id").alias("avg_id")
    )
    .orderBy("bucket")
)


df.show()
```

```python
try:
    catalogs = spark.sql("SHOW CATALOGS").limit(5).collect()
    for cat in catalogs:
        print(f" • {cat[0]}")
except Exception as e:
    print(f"Exception: {e}")
```
