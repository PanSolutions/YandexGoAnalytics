## Prerequisites & execution order

**1 · Configure the environment** — create a `.env` file in the project root:

```dotenv
DATABRICKS_HOST=https://<your-workspace>.cloud.databricks.com
DATABRICKS_CATALOG=yandex_go_dev
DATABRICKS_ENV=dev
```

**2 · Authenticate and deploy the infrastructure**

```bash
databricks auth login --host https://<your-workspace>.cloud.databricks.com
make validate      # databricks bundle validate -t dev
make deploy-dev    # schemas, Volume, DLT pipeline, scheduled job
```

**3 · Run the notebooks `1 → 8`**

| Mode                                    | How it connects |
|:----------------------------------------|:--|
| **Local** (PyCharm / VS Code / Jupyter) | `databricks-connect` opens a **serverless** session |
| **Databricks workspace**                | The active Spark session is reused automatically |

> **Heads-up** — the generators *append* files, and Auto Loader remembers what it has already processed.
> Re-running a notebook is safe, but it will add more mock data on top of the existing data.
